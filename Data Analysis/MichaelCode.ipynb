```python
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import numpy as np

df = pd.read_csv('../data/claims_train.csv')

df['exposure*bonusmalus'] = df['Exposure'] * df['BonusMalus']

df['Density_over_VehAge'] = df['Density'] / (df['VehAge'] + 1e-6)


X = df[['Density', 'DrivAge', 'Exposure', 'VehAge','VehPower', 'BonusMalus', 'VehGas', 'exposure*bonusmalus', 'Density_over_VehAge', 'VehBrand']]
X = pd.get_dummies(X, drop_first=True)
y = df['ClaimNb']

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42)
clf = RandomForestClassifier(random_state=42)
clf.fit(x_train, y_train)
y_hat = clf.predict(x_test)

print(classification_report(y_test,y_hat))
```
